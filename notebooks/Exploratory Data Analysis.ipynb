```python
print(f"Skewness: {data['Weight (Kg)'].skew():.2f}")
```

```python
# Before:

fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15,4))

sns.kdeplot(data=data, x='Weight (Kg)', fill=True, ax=ax[0])
ax[0].set_title("Distribution Plot")

sns.boxplot(data=data, y='Weight (Kg)', ax=ax[1])
ax[1].set_title("Box Plot")

sns.regplot(data=data, x='Weight (Kg)', y='Price (Rs)',scatter_kws={'color' : 'blue', 'alpha' : 0.5}, line_kws={'color' : 'red', 'linewidth' : 3}, ax=ax[2])
ax[2].set_title("Weight vs Price")

plt.tight_layout()
plt.show()
```

```python
# After:

test_data = data.copy()

transform_weight = FunctionTransformer(func=np.log1p)
test_data['Weight (Kg)'] = transform_weight.fit_transform(test_data[['Weight (Kg)']])


fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15,4))

sns.kdeplot(data=test_data, x='Weight (Kg)', fill=True, ax=ax[0])
ax[0].set_title("Distribution Plot")

sns.boxplot(data=test_data, y='Weight (Kg)', ax=ax[1])
ax[1].set_title("Box Plot")

sns.regplot(data=test_data, x='Weight (Kg)', y='Price (Rs)',scatter_kws={'color' : 'blue', 'alpha' : 0.5}, line_kws={'color' : 'red', 'linewidth' : 3}, ax=ax[2])
ax[2].set_title("Transformed Weight vs Price")

plt.tight_layout()
plt.show()
```

```python
test_data[['Weight (Kg)', 'Price (Rs)']].corr()
```

#### **Weight (Kg) -> Simple-Imputation(median).**

- After *Log-Transformation*, there is no such improvement in linearity/correaltion, therefore the orginal feature will be retained.

- Feature is right-skewed, therefore imputing missing values with *median*.
